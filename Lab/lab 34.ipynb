```python
import unittest

class SimpleInterestTest:

    # Method to calculate simple interest based on customer type
    def calculate_interest(self, principal: float, rate: float, time: int, customer_type: str) -> float:
        if customer_type.lower() == "senior":
            rate = rate - 2
        elif customer_type.lower() == "regular":
            rate = rate
        else:
            raise ValueError("Invalid Customer Type")

        return (principal * rate * time) / 100

# Python equivalent of JUnit test cases
class TestSimpleInterest(unittest.TestCase):

    def setUp(self):
        self.obj = SimpleInterestTest()

    def test_senior_customer(self):
        self.assertAlmostEqual(1600.0, self.obj.calculate_interest(10000, 10, 2, "Senior"), delta=0.01)

    def test_regular_customer(self):
        self.assertAlmostEqual(2000.0, self.obj.calculate_interest(10000, 10, 2, "Regular"), delta=0.01)

    def test_invalid_customer(self):
        with self.assertRaises(ValueError):
            self.obj.calculate_interest(10000, 10, 2, "Student")

if __name__ == '__main__':
    # Demonstration running similar to main method
    obj = SimpleInterestTest()
    interest = obj.calculate_interest(10000, 10, 2, "Senior")
    print("Customer Type: Senior")
    print("Principal Amount: 10000")
    print(f"Simple Interest: {interest}\n")

    # Run the tests
    suite = unittest.TestLoader().loadTestsFromTestCase(TestSimpleInterest)
    unittest.TextTestRunner(verbosity=2).run(suite)
```
