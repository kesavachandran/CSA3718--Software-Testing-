```python
import unittest

class VotingEligibility:
    def check_eligibility(self, age: int) -> str:
        if age >= 18:
            return "Eligible to Vote"
        else:
            return "Not Eligible to Vote"

class TestVotingEligibility(unittest.TestCase):
    def setUp(self):
        self.validator = VotingEligibility()

    def test_eligible_age(self):
        self.assertEqual(self.validator.check_eligibility(20), "Eligible to Vote")
        self.assertEqual(self.validator.check_eligibility(18), "Eligible to Vote")

    def test_not_eligible_age(self):
        self.assertEqual(self.validator.check_eligibility(17), "Not Eligible to Vote")
        self.assertEqual(self.validator.check_eligibility(0), "Not Eligible to Vote")

# Run the tests in the notebook
suite = unittest.TestLoader().loadTestsFromTestCase(TestVotingEligibility)
unittest.TextTestRunner(verbosity=2).run(suite)
```
