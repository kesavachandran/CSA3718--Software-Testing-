```python
import unittest

class PalindromeTest:

    # Method to check palindrome number
    def is_palindrome(self, number: int) -> bool:
        if number < 0:
            return False

        original = number
        reverse = 0
        temp = number

        while temp > 0:
            digit = temp % 10
            reverse = reverse * 10 + digit
            temp = temp // 10

        return original == reverse

# Python equivalent of the JUnit Test Cases
class TestPalindrome(unittest.TestCase):

    def setUp(self):
        self.obj = PalindromeTest()

    def test_palindrome_number(self):
        self.assertTrue(self.obj.is_palindrome(121))

    def test_non_palindrome_number(self):
        self.assertFalse(self.obj.is_palindrome(123))

    def test_single_digit_number(self):
        self.assertTrue(self.obj.is_palindrome(7))

    def test_negative_number(self):
        self.assertFalse(self.obj.is_palindrome(-121))

    def test_zero(self):
        self.assertTrue(self.obj.is_palindrome(0))

if __name__ == '__main__':
    # Demonstration running similar to main method
    obj = PalindromeTest()
    number = 121
    print(f"Given Number: {number}")
    if obj.is_palindrome(number):
        print("Palindrome Number\n")
    else:
        print("Not a Palindrome Number\n")

    # Running the assertions
    suite = unittest.TestLoader().loadTestsFromTestCase(TestPalindrome)
    unittest.TextTestRunner(verbosity=2).run(suite)
```
