### 32. String Comparison and Validation using Unit Test Assertions
We will use Python's `unittest` framework to demonstrate standard assertions like `assertEqual` and `assertTrue` for validating string processing.

```python
import unittest

def format_message(name):
    return f"Hello, {name}!"

class TestStringMethods(unittest.TestCase):
    def test_string_equality(self):
        # Validate that the formatted output matches the expected greeting
        result = format_message("Alice")
        self.assertEqual(result, "Hello, Alice!")

    def test_string_contains(self):
        # Validate that the output contains the name
        result = format_message("Bob")
        self.assertIn("Bob", result)

# Run the tests directly in the notebook
suite = unittest.TestLoader().loadTestsFromTestCase(TestStringMethods)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
```
